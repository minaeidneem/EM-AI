```python
import cv2
import numpy as np
import pickle
import face_recognition
from ultralytics import YOLO

# 1. تحميل الموديلات المحدثة (النسخة الصغيرة s للدقة العالية)
print("⌛ جاري تحميل الموديلات الاحترافية وقاعدة بيانات الوجوه...")
model_pose = YOLO('yolov8s-pose.pt')  # نسخة s لزيادة دقة كشف الوضعية والنوم
model_obj = YOLO('yolov8s.pt')        # نسخة s لزيادة دقة كشف الهاتف والأشخاص

# تحميل قاعدة بيانات الوجوه
try:
    with open("trained_faces_data.pkl", "rb") as f:
        face_data = pickle.load(f)
except FileNotFoundError:
    print("❌ خطأ: ملف trained_faces_data.pkl غير موجود!")
    face_data = {"encodings": [], "names": []}

PHONE_ID = 67
cap = cv2.VideoCapture(0)

# قوائم لتخزين الأرقام وحساب المتوسط النهائي
all_face_accuracies = []
all_phone_confidences = []

print("🚀 EM-AI High-Accuracy System Online... اضغط 'q' لإغلاق الكاميرا وحساب الدقة الجديدة.")

while cap.isOpened():
    success, frame = cap.read()
    if not success: 
        break
        
    h, w, _ = frame.shape
    annotated_frame = frame.copy()
    
    current_face_acc = 0.0
    current_phone_conf = 0.0
    
    # ----------------------------------------------------
    # 1. التعرف على الوجوه بحجم صورة أكبر ومعادلة دقة محسنة
    # ----------------------------------------------------
    small_frame = cv2.resize(frame, (0, 0), fx=0.5, fy=0.5)
    rgb_small_frame = cv2.cvtColor(small_frame, cv2.COLOR_BGR2RGB)

    face_locations = face_recognition.face_locations(rgb_small_frame)
    face_encodings = face_recognition.face_encodings(rgb_small_frame, face_locations)

    face_names = []
    for face_encoding in face_encodings:
        matches = face_recognition.compare_faces(face_data["encodings"], face_encoding, tolerance=0.55)
        name = "Unknown"

        face_distances = face_recognition.face_distance(face_data["encodings"], face_encoding)
        if len(face_distances) > 0:
            best_match_index = np.argmin(face_distances)
            if matches[best_match_index]:
                name = face_data["names"][best_match_index]
                
                distance = face_distances[best_match_index]
                if distance < 0.5:
                    current_face_acc = 90.0 + (0.5 - distance) * 20  
                else:
                    current_face_acc = (1.0 - distance) * 100
                
                current_face_acc = max(0.0, min(100.0, current_face_acc))

        face_names.append(name)
        # تعديل: حفظ الدقة في القائمة فوراً لكل وجه معروف تم رصده في هذا الإطار
        if name != "Unknown" and current_face_acc > 0:
            all_face_accuracies.append(current_face_acc)

    current_employee = face_names[0] if face_names else "Active Person"

    # ----------------------------------------------------
    # 2. كشف الأشياء بالاعتماد على YOLOv8s
    # ----------------------------------------------------
    obj_results = model_obj(frame, conf=0.35, verbose=False) 
    detected_objects = []
    phone_box = None
    
    for r in obj_results:
        annotated_frame = r.plot()  
        detected_objects = [r.names[int(c)] for c in r.boxes.cls]
        
        for box in r.boxes:
            cls = int(box.cls[0])
            if cls == PHONE_ID:
                phone_box = list(map(int, box.xyxy[0]))
                current_phone_conf = float(box.conf[0]) * 100
                all_phone_confidences.append(current_phone_conf)

    # ----------------------------------------------------
    # 3. منطق مراقبة الحالة (Pose)
    # ----------------------------------------------------
    status = "Status: Desk is Empty"
    color = (255, 0, 0) 

    if 'person' in detected_objects:
        status = f"Employee: {current_employee} ({current_face_acc:.1f}%)"
        color = (0, 255, 0) 
        
        if 'cell phone' in detected_objects:
            status = f"Warning: Distracted (Phone Conf: {current_phone_conf:.1f}%)"
            color = (0, 165, 255) 
            
        pose_results = model_pose(frame, verbose=False)
        for r in pose_results:
            if r.keypoints is not None and len(r.keypoints.data) > 0:
                kpts = r.keypoints.data[0].cpu().numpy()
                nose = kpts[0]; l_ear = kpts[3]; l_shoulder = kpts[5]; r_shoulder = kpts[6]
                avg_shoulder_y = (l_shoulder[1] + r_shoulder[1]) / 2

                if phone_box is not None:
                    px1, py1, px2, py2 = phone_box
                    ear_x, ear_y = l_ear[0], l_ear[1]
                    if (px1 - 40 < ear_x < px2 + 40) and (py1 - 40 < ear_y < py2 + 40):
                        status = f"CRITICAL: {current_employee.upper()} TALKING ON PHONE!"
                        color = (0, 0, 255) 

                if nose[2] > 0.3 and l_shoulder[2] > 0.3:
                    if (avg_shoulder_y - nose[1]) < 45: 
                        status = f"CRITICAL: {current_employee.upper()} SLEEPING!"
                        color = (0, 0, 255) 

    # ----------------------------------------------------
    # 4. رسم مربعات الوجوه وضبط الأبعاد (ضرب 2 لأن الحجم 0.5)
    # ----------------------------------------------------
    for (top, right, bottom, left), name in zip(face_locations, face_names):
        top *= 2; right *= 2; bottom *= 2; left *= 2  
        cv2.rectangle(annotated_frame, (left, top), (right, bottom), (0, 255, 0), 2)
        cv2.rectangle(annotated_frame, (left, bottom - 35), (right, bottom), (0, 255, 0), cv2.FILLED)
        cv2.putText(annotated_frame, f"{name} {current_face_acc:.1f}%", (left + 6, bottom - 6), 
                    cv2.FONT_HERSHEY_DUPLEX, 0.5, (255, 255, 255), 1)

    # شريط الحالة العلوي
    cv2.rectangle(annotated_frame, (0, 0), (w, 60), (0, 0, 0), -1)
    cv2.putText(annotated_frame, status, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.75, color, 2, cv2.LINE_AA)

    cv2.imshow('GuardIQ - High Accuracy Mode', annotated_frame)
    
    # تحسين استجابة زر الـ Q لتفادي تعليق OpenCV
    key = cv2.waitKey(1) & 0xFF
    if key == ord('q') or key == ord('Q'): 
        break

# إغلاق آمن ومضمون للنوافذ قبل الطباعة
cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1) # سطر إضافي لضمان مسح النوافذ من ذاكرة نظام التشغيل فوراً

# ----------------------------------------------------
# 5. طباعة التقرير النهائي عالي الدقة
# ----------------------------------------------------
print("\n" + "="*40)
print("📊 تقرير دقة نظام EM-AI المطور والنهائي:")
print("="*40)

if all_face_accuracies:
    avg_face_acc = np.mean(all_face_accuracies)
    print(f"🎯 متوسط دقة التعرف على الوجوه الإجمالية: {avg_face_acc:.2f}%")
else:
    print("🎯 متوسط دقة التعرف على الوجوه: لم يتم جمع بيانات كافية (تأكد من ظهور الوجه بوضوح).")

if all_phone_confidences:
    avg_phone_conf = np.mean(all_phone_confidences)
    print(f"📱 متوسط ثقة الموديل في كشف الهاتف: {avg_phone_conf:.2f}%")
else:
    print("📱 متوسط ثقة الموديل في كشف الهاتف: لم يتم رصد هواتف.")
print("="*40)
```
